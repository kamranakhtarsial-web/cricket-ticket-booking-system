```python
def show_menu():
    print("===== CRICKET TICKET SYSTEM =====")
    print("1. Display All Tickets")
    print("2. Search Ticket")
    print("3. Book Ticket")
    print("4. Display Available Tickets")
    print("5. Display Sold Tickets")
    print("6. Remaining Tickets")
    print("7. View Seat Map")
    print("8. Exit")
```

```python
def find_ticket(seat):
    if seat in tickets:
        return tickets[seat]

    return None
```

```python
def display_all():
    print("----- ALL TICKETS -----")

    for seat in tickets:
        ticket = tickets[seat]

        print("Seat:", ticket.seat,
              "| Category:", ticket.category,
              "| Price:", ticket.price,
              "| Status:", ticket.status)
```

```python
def search_ticket():
    seat = input("Enter seat number: ")

    if seat.isdigit():
        seat = int(seat)
        ticket = find_ticket(seat)

        if ticket == None:
            print("Invalid seat number!")

        else:
            print("Seat:", ticket.seat)
            print("Category:", ticket.category)
            print("Price:", ticket.price)
            print("Status:", ticket.status)

    else:
        print("Please enter a valid number.")
```

```python
def book_ticket():
    print("--- Book Tickets ---")

    quantity = int(input("How many tickets do you want to book? "))

    count = 0

    while count < quantity:
        seat = int(input("Enter seat number: "))

        if seat in tickets:
            if tickets[seat].status == "Available":
                tickets[seat].book()

                print("Ticket booked successfully!")
                print("Seat Number:", seat)
                print("Category:", tickets[seat].category)
                print("Price: Rs.", tickets[seat].price)

                count = count + 1

            else:
                print("Sorry! This seat is already sold.")

        else:
            print("Invalid seat number!")

    print("Booking process completed.")
```

```python
def display_by_status(status):
    total = 0

    for seat in tickets:
        ticket = tickets[seat]

        if ticket.status == status:
            print("Seat:", ticket.seat,
                  "| Category:", ticket.category,
                  "| Price:", ticket.price)

            total += 1

    print("Total", status, "tickets:", total)


def display_available():
    print("----- AVAILABLE TICKETS -----")
    display_by_status("Available")


def display_sold():
    print("----- SOLD TICKETS -----")
    display_by_status("Sold")
```

```python
def remaining_tickets():
    total = 0
    sold = 0
    available = 0
    revenue = 0

    for seat in tickets:
        total += 1

        if tickets[seat].status == "Sold":
            sold += 1
            revenue += tickets[seat].price

        else:
            available += 1

    percentage = (sold / total) * 100

    print("----- TICKET SUMMARY -----")
    print("Total Tickets:", total)
    print("Sold Tickets:", sold)
    print("Available Tickets:", available)
    print("Percentage Sold:", percentage, "%")
    print("Total Money Earned: Rs.", revenue)
```

```python
def seat_map():
    print("----- SEAT MAP -----")
    count = 0

    for seat in tickets:

        if tickets[seat].status == "Sold":
            print("[XXX]", end=" ")

        else:
            print("[", seat, "]", end=" ")

        count += 1

        if count == 10:
            print()
            count = 0
```

```python
while True:

    show_menu()

    choice = input("Enter your choice: ")

    if choice == "1":
        display_all()

    elif choice == "2":
        search_ticket()

    elif choice == "3":
        book_ticket()

    elif choice == "4":
        display_available()

    elif choice == "5":
        display_sold()

    elif choice == "6":
        remaining_tickets()

    elif choice == "7":
        seat_map()

    elif choice == "8":
        print("Thank you for using Cricket Ticket System!")
        break

    else:
        print("Invalid choice! Please enter 1 to 8.")
```
